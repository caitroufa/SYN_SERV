```python
#function to convert water year to calendar year and output daily count
def Cumu_sum(df):

    daily_counts = pd.DataFrame(index=np.arange(1,366,1))
    df['Day_of_Year'] = pd.to_datetime(df['Date']).dt.dayofyear
    
    for year in df['Year'].unique()[:-1]:
        b=[]
        
        if year == np.min(df['Year'].unique()):
            df_year = df[((df['Year'] == np.min(df['Year'].unique())) | (df['Year'] == np.max(df['Year'].unique())))]
            
            for day in np.arange(1,366,1):
                df_sub = df_year[(df_year['Day_of_Year'] == day)]
                b.append(len(df_sub))
            
            
        else:
            df_year = df[(df['Year'] == year)]
            
            for day in np.arange(1,366,1):
                df_sub = df_year[(df_year['Day_of_Year'] == day)]
                b.append(len(df_sub))
            
        daily_counts[year] = b
            
    return daily_counts
```

```python
#calculate daily counts for each epoch and convert to calendar year
new_df_HIST = Cumu_sum(df_HIST)
new_df_4p5MID = Cumu_sum(df_4p5MID)
new_df_8p5MID = Cumu_sum(df_8p5MID)
new_df_4p5END = Cumu_sum(df_4p5END)
new_df_8p5END = Cumu_sum(df_8p5END)
```

```python
# function to calc severe days and convert from water year to calendar year
def Cumu_days(df):
    daily_counts = pd.DataFrame(index=np.arange(1, 366, 1))
    df['Day_of_Year'] = pd.to_datetime(df['Date']).dt.dayofyear
    
    for year in df['Year'].unique()[:-1]:
        b = []
        
        if year == np.min(df['Year'].unique()):
            df_year = df[((df['Year'] == np.min(df['Year'].unique())) | (df['Year'] == np.max(df['Year'].unique())))]
            
            for day in np.arange(1, 366, 1):
                # if severe on a day, count as 1, otherwise 0
                df_sub = df_year[(df_year['Day_of_Year'] == day)]
                b.append(1 if len(df_sub) > 0 else 0)
                
        else:
            df_year = df[(df['Year'] == year)]
            
            for day in np.arange(1, 366, 1):
                df_sub = df_year[(df_year['Day_of_Year'] == day)]
                b.append(1 if len(df_sub) > 0 else 0)
            
        daily_counts[year] = b
            
    return daily_counts
```

```python
# calculate for each epoch using new days function
new_df_HIST_days = Cumu_days(df_HIST)
new_df_4p5MID_days = Cumu_days(df_4p5MID)
new_df_8p5MID_days = Cumu_days(df_8p5MID)
new_df_4p5END_days = Cumu_days(df_4p5END)
new_df_8p5END_days = Cumu_days(df_8p5END)
```

```python
# Define day-of-year ranges for each season
seasons = {
    'DJF': np.concatenate([np.arange(0, 59), np.arange(334, 365)]),  # days 1–59 and 335–365 (0-based)
    'MAM': np.arange(59, 151),   # Mar 1 – May 31
    'JJA': np.arange(151, 243),  # Jun 1 – Aug 31
    'SON': np.arange(243, 334)   # Sep 1 – Nov 30
}

# Function to compute slope of cumulative frequency
def get_slope(x, y):
    slope, _ = np.polyfit(x, y, 1)
    return slope  # units: events/day

# Store cumulative series for each epoch
cumu_series = {
    'HIST': np.cumsum(new_df_HIST.mean(axis=1)).values,
    'MID4.5': np.cumsum(new_df_4p5MID.mean(axis=1)).values,
    'MID8.5': np.cumsum(new_df_8p5MID.mean(axis=1)).values,
    'END4.5': np.cumsum(new_df_4p5END.mean(axis=1)).values,
    'END8.5': np.cumsum(new_df_8p5END.mean(axis=1)).values
}

# Compute full-year and seasonal slopes
slope_results = {}

for label, series in cumu_series.items():
    results = {}
    # Full year slope
    x_full = np.arange(365)
    y_full = series
    results['FullYear'] = get_slope(x_full, y_full)

    # Seasonal slopes
    for season, days in seasons.items():
        x = days
        y = series[days]
        results[season] = get_slope(x, y)
    
    slope_results[label] = results

# Convert to DataFrame for display
slope_df = pd.DataFrame(slope_results).T  # epochs as rows
print("Slope of Cumulative Frequency Curves (events/day):")
print(slope_df.round(4))
```

```python
# Bar plot of normalized slopes
seasons_to_plot = ['DJF', 'MAM', 'JJA', 'SON']
epochs = list(slope_df.index)

x = np.arange(len(seasons_to_plot))  # positions for DJF, MAM, JJA, SON
bar_width = 0.15

fig, ax = plt.subplots(figsize=(8, 5), dpi=150)

colors = {
    'HIST': '#c7e9c0',
    'MID4.5': '#6baed6',
    'MID8.5': '#a50f15',
    'END4.5': '#4292c6',
    'END8.5': '#67000d'
}

# Plot each epoch as a set of bars
for i, epoch in enumerate(epochs):
    offset = (i - 2) * bar_width  # centers around zero
    ax.bar(x + offset, slope_df.loc[epoch, seasons_to_plot], width=bar_width,
           label=epoch, color=colors.get(epoch, 'gray'))

# Formatting
ax.set_xticks(x)
ax.set_xticklabels(seasons_to_plot, fontsize=12)
ax.set_ylabel("Slope (events/day)", fontsize=12)
ax.set_title("Seasonal Slopes of Cumulative Event Frequency", fontsize=14)
ax.grid(axis='y', alpha=0.3)
ax.legend(title="Epoch")

plt.tight_layout()
plt.show()
```
